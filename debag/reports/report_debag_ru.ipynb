## Назначение

Этот набор не проверяет корректность основных алгоритмов. Он проверяет саму методику бенчмарка: три контрольных модуля запускают одну и ту же детерминированную нагрузку под разными именами. Поэтому значимое расхождение их результатов означает влияние системного шума, порядка запуска, прогрева кэшей или доступности данных.

Раннер чередует порядок алгоритмов, прогревает поиск, временно отключает сборщик мусора и использует медиану центральных измерений. После запуска `verify_results.py` сравнивает медианный относительный разброс с порогом 5%.

## Запуск

```bash
./scripts/run_debag.sh
```

Команда временно создаёт CSV, проверяет допустимый разброс, обновляет отчёт и общие графики, затем удаляет промежуточные данные.

## Выводы

Линии трёх контролей должны оставаться близкими. Автоматическая проверка считается пройденной, если медианный относительный разброс времени подготовки и поиска не превышает 5%. Это не гарантирует полное отсутствие шума, но выявляет систематическое преимущество из-за порядка запуска и нестабильность среды, достаточную для искажения сравнения основных алгоритмов.
